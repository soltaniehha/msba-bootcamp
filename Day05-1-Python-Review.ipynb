## Part 8: Something New: Lambda Functions

You have written functions with `def` all week. A **lambda function** is a one-line function for cases where `def` feels like too much ceremony. These two definitions do exactly the same thing:

```python
def add_tip(amount):
    return round(amount * 1.18, 2)

add_tip = lambda amount: round(amount * 1.18, 2)
```

Read the lambda as "take `amount`, return `round(amount * 1.18, 2)`". Whatever comes after the colon is what gets returned; no `return` keyword needed. You will meet lambdas again soon: pandas leans on them for quick one-off transformations.

## Rapid-Fire Reminders

A quick checklist before you put this week to work:

- Pseudocode is a plain-English outline of an algorithm's steps; Python is an interpreted language, running your code one line at a time
- `input()` always returns a string; wrap it in `int()` or `float()` before doing math (adding a string to a number raises a TypeError)
- `xs[1:4]` starts at index 1 and stops before index 4; negative indexes count from the end (`xs[-1]` is the last item)
- `range(5)` produces 0 through 4; `range(1, n)` starts at 1 and stops before `n`
- A `while` loop runs as long as its condition is True; something inside it has to change, or it never ends
- A set keeps one copy of each value, so `set()` on a list drops the duplicates
- A function with no `return` statement returns `None`
- `return a, b` returns a tuple you can unpack into two names
- Parameters with defaults (`rate=0.0625`) are optional when calling
- A class is the blueprint; each object is an instance of it; `self` is the object a method was called on
- `import helpers` then `helpers.clean(...)`, or `from helpers import clean` then just `clean(...)`; fix the module once and everything that imports it gets the fix
- `except ValueError:` catches only that error type, which beats a bare `except:`
- Add `?` after any object (try `len?` in a cell) to see its documentation
- `git clone` downloads a repo to your computer; `git add`, then `git commit`, then `git push`: push is the one that uploads your commits to GitHub

You have covered a lot of ground in four days. Trust your notebooks, and good luck.
