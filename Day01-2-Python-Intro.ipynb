## Part 2: Pseudocode

### What is Pseudocode?
A bridge between human thinking and computer code, written in plain English but structured like programming logic.

### Key Elements:
- `START` and `END`
- `INPUT` and `OUTPUT`
- `IF`, `THEN`, `ELSE`
- `WHILE`, `FOR`
- Indentation for structure

### Example: Checking if someone can vote
```
START
    INPUT person's age
    IF age >= 18 THEN
        OUTPUT "You can vote!"
    ELSE
        OUTPUT "Not old enough yet"
    END IF
END
```
