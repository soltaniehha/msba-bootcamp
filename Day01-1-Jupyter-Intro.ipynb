## Part 1: What Is a Notebook?

A notebook is a document that mixes explanation and runnable code. You read a little, run a little, see the result right away, and keep going.

The file you are looking at ends in `.ipynb`, short for "interactive Python notebook". People who work with data like notebooks because analysis is a back-and-forth process: try something, look at the output, adjust. A notebook keeps that whole process in one file.

---

- **Code cells** hold Python that the computer executes.

## Part 3: Running Cells

Click a code cell and press **Shift+Enter** to run it and move to the next cell. **Ctrl+Enter** runs it and stays put. There is also a **Run All** button at the top that runs every cell from top to bottom.

The number in brackets next to a code cell, like `[1]`, shows the order in which cells were run. Run the cell below a few times and watch the number climb.

````python
print("You ran this cell.")
````

---

## Part 4: Cells Share Memory

Code cells are not isolated. A variable defined in one cell is available in every cell you run after it. Run the next two cells in order.

````python
monthly_sales = 2500
````

````python
monthly_sales * 12
````

The second cell knew about `monthly_sales` even though it was defined in a different cell. That is what makes a notebook feel like one continuous program.

It also means **order matters**. Running cells out of order is the most common source of confusion in notebooks. If your results stop making sense, use **Restart and Run All**: it wipes the slate clean and runs everything from the top.

---

## Part 5: The Kernel

Behind every notebook sits a **kernel**, the Python process that actually runs your code and remembers your variables. Restarting the kernel erases that memory and gives you a fresh start. When a notebook gets stuck or behaves strangely, restarting the kernel is the standard fix. Your code and text stay in the file; only the variables are forgotten.

## Part 7: Notebooks in VS Code

VS Code is what we'll use in class. A few things to know:

- **Opening a notebook**: double-click any `.ipynb` file and it opens in the notebook editor.
- **Picking a kernel**: click **Select Kernel** in the top right and choose your Python installation. You do this once per notebook. If code cells refuse to run, a missing kernel is usually the reason.
- **Running cells**: each cell has a run arrow to its left, and Shift+Enter works as usual.
- **Adding cells**: hover between cells and click **+ Code** or **+ Markdown**.
- **Saving**: Cmd+S on a Mac, Ctrl+S on Windows.

We'll do all of this together, so don't worry about memorizing it now.
