## Part 1: Python Files (.py)

So far all of our code has lived in notebook cells. Out in the wild, most Python code lives in plain text files ending in `.py`, usually called scripts. Notebooks are great for exploring; `.py` files are how code gets shared, scheduled, and shipped.

The `%%writefile` magic at the top of a cell saves the cell's contents to a file instead of running it. Let's create a tiny script:

````python
%%writefile greet_customer.py
# A tiny Python script
name = "Ava"
print(f"Welcome back, {name}!")
print("You have 2 new offers today.")
````

### Running a .py file from the command line

Open a terminal (in VS Code: View, then Terminal), make sure you are in this folder, and type:

```
python greet_customer.py
```

Python reads the file and runs it top to bottom, just like running notebook cells in order.

### Running a .py file inside a notebook

Notebooks come with **magic functions**: special commands that start with `%` (line magics, acting on one line) or `%%` (cell magics, acting on the whole cell). They are not Python itself; they are notebook extras that make common chores easy. `%%writefile` above was one of them.

The `%run` magic executes a `.py` file right inside the notebook:

````python
# Run the script with the %run magic function
%run greet_customer.py
````

A related trick: starting a line with `!` hands it to the terminal, so `!python greet_customer.py` does the same job through the shell. We'll use that form later to run a file exactly the way the command line does.

## Part 2: Creating a Simple Module
We'll sketch a module `analytics_utils.py` with reusable functions.

````python
%%writefile analytics_utils.py
def summarize_sales(sales):
    total = sum(sales)
    avg = round(total / len(sales), 2) if sales else 0
    return {"total": total, "average": avg}

def apply_tax(amount, rate=0.0625):
    return round(amount * (1 + rate), 2)
````

## Part 3: Import and Use the Module

````python
# Live Code: Import usage (skeleton)

from analytics_utils import summarize_sales, apply_tax
sales = [120, 98, 145]

summary = summarize_sales(sales)
taxed = apply_tax(100)
summary, taxed
````

## Part 4: Scripts vs. Modules: the `__main__` Guard

The same `.py` file can be run directly or imported. Python sets a special variable `__name__` to `"__main__"` when the file is run directly, and to the module's name when it is imported. A guard lets you attach demo or self-test code that runs only in the first case:

````python
%%writefile report_tools.py
def format_currency(amount):
    return f"${amount:,.2f}"

if __name__ == "__main__":
    # This line runs only when the file is executed directly
    print("Self-test:", format_currency(1234.5))
````

````python
# Run directly: the self-test fires
!python report_tools.py
````

````python
# Imported: the self-test stays quiet
from report_tools import format_currency
format_currency(89.5)
````
